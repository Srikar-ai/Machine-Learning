```python
import numpy as np 
```

```python
class NaiveBayes:

    def __init__(self,smoothing):
        self.smoothing = smoothing

    def fit(self,X,y):

        classes,count = np.unique(y,return_counts=True)

        self.classes = classes

        self.priors = {}
        self.feature_probability = {}

        for i in classes:
            X_c = X[y == i]
            self.priors[i] = len(X_c)/X.shape[0]
            self.feature_probability[i] = (
                    np.sum(X_c, axis=0) + self.smoothing
                ) / (
                    len(X_c) + 2 * self.smoothing
                )

    def predict(self,X):

        predictions = []

        for x in X:

            class_scores = []

            for c in self.classes:

                score = np.log(self.priors[c])

                p = self.feature_probability[c]

                score += np.sum(
                    x * np.log(p) + (1-x)*np.log(1-p)
                )
                class_scores.append(score)

            predictions.append(
                self.classes[np.argmax(class_scores)]
            )
        return predictions
```

```python
X = np.array([
    [1, 1, 0],
    [1, 1, 1],
    [0, 0, 1],
    [0, 0, 0]
])

y = np.array([1, 1, 0, 0])

model = NaiveBayes(smoothing=1)
model.fit(X, y)

predictions = model.predict(X)

print("Predictions:", predictions)
print("Actual:", y)
print("Accuracy:", np.mean(predictions == y))
```
