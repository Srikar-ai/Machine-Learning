```python
import  numpy as np 
```

```python
class NaiveBayes:

    def __init__(self,smoothing):
        self.smoothing = smoothing

    def fit(self,X,y):

        classes,count = np.unique(y,return_counts=True)

        self.classes = classes

        self.priors = {}
        self.mean = {}
        self.variance = {}

        for i in classes:
            X_c = X[y == i]
            self.priors[i] = len(X_c)/X.shape[0]
            self.mean[i] = np.mean(X_c,axis=0)
            self.variance[i] = np.var(X_c,axis=0)


    def predict(self,X):

        predictions = []

        for x in X:

            class_scores = []

            for c in self.classes:

                mu = self.mean[c]
                var = np.maximum(self.variance[c], 1e-9)

                score = np.log(self.priors[c])

                # p = self.feature_probability[c]

                score += np.sum(
                -0.5 * np.log(2 * np.pi * var)
                - (x - mu) ** 2 / (2 * var)
                )
                class_scores.append(score)

            predictions.append(
                self.classes[np.argmax(class_scores)]
            )
        return predictions
```

```python
# Three classes, two continuous features
X_train = np.array([
    [1.0, 2.0],
    [1.2, 1.8],
    [0.8, 2.2],
    [5.0, 6.0],
    [5.2, 5.8],
    [4.8, 6.2],
    [9.0, 1.0],
    [9.2, 1.2],
    [8.8, 0.8]
])

y_train = np.array([
    0, 0, 0,
    1, 1, 1,
    2, 2, 2
])

X_test = np.array([
    [1.1, 2.1],
    [5.1, 6.1],
    [9.1, 1.1]
])

y_test = np.array([0, 1, 2])

model = NaiveBayes(smoothing=1e-9)
model.fit(X_train, y_train)

predictions = model.predict(X_test)

print("Actual:     ", y_test)
print("Predictions:", predictions)
print("Accuracy:", np.mean(predictions == y_test))
```
