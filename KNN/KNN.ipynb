```python
import numpy as np 
```

```python
class KNN:

    def __init__(self,n,X,y):
        self.neighbours = n
        self.X = X
        self.y = y 

    def find_cluster(self,X_test):

        predictions = []

        for x in X_test:

            k_distances = []
            for i in range(len(self.X)):
                # Calculate the distance 
                distance = np.sqrt(np.sum((x - self.X[i]) ** 2))
                k_distances.append((distance, self.y[i]))

            k_distances.sort(key=lambda item: item[0])

            nearest_labels = [
                label for distance, label
                in k_distances[:self.neighbours]
            ]
            classes, counts = np.unique(
                nearest_labels, return_counts=True
            )
            predictions.append(classes[np.argmax(counts)])
        return predictions 
```

```python
import numpy as np

X = np.array([
    [1, 2],
    [2, 3],
    [3, 3],
    [8, 8],
    [9, 8]
])

y = np.array([0, 0, 0, 1, 1])

X_test = np.array([
    [2, 2],
    [8, 9]
])

knn = KNN(n=3, X=X, y=y)
print(knn.find_cluster(X_test))
```
