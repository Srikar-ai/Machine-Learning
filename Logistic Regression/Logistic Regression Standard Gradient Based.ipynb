```python
import numpy as np 
import pandas as pd 
from time import time 
```

```python
class LogisticRegression:

    def __init__(self,epochs,input_features,output_classes,lr):
        self.lr = lr 
        self.weights = np.random.rand(input_features,output_classes)
        self.bias = np.zeros(1)
        self.epochs = epochs 

    def forward(self,X):
        return X @ self.weights + self.bias

    def sigmoid(self,Z):
        return (1/(1+np.exp(-Z)))

    def softmax(self,Z):
        exp_Z = np.exp(Z - np.max(Z, axis=1, keepdims=True))
        return exp_Z / np.sum(exp_Z, axis=1, keepdims=True)

    def loss(self,y,y_pred):
        return -np.mean(np.log(y_pred[np.arange(len(y)),y]))

    def fit(self,X,y):


        for epoch in range(self.epochs):

            Z = self.forward(X)
            probabilites = self.softmax(Z)
            loss = self.loss(y,probabilites)

            DL_DZ = probabilites.copy()

            DL_DZ[np.arange(X.shape[0]), y] -= 1

            DL_DZ /= X.shape[0]
            DL_DW = X.T @ DL_DZ 
            DL_DB = np.sum(DL_DZ,axis=0)


            self.weights  = self.weights - self.lr * DL_DW 
            self.bias = self.bias - self.lr * DL_DB

            if epoch%100 == 0:
                print("Loss : ",loss)
```

```python
np.random.seed(42)

# Class 0
X0 = np.random.randn(100, 2) + np.array([-2, -2])

# Class 1
X1 = np.random.randn(100, 2) + np.array([2, -2])

# Class 2
X2 = np.random.randn(100, 2) + np.array([0, 2])

# Combine
X = np.vstack([X0, X1, X2])

# Labels
y = np.array(
    [0] * 100 +
    [1] * 100 +
    [2] * 100
)

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Classes:", np.unique(y))
```

```python
start = time()

model = LogisticRegression(
    epochs=5000,
    input_features=2,
    output_classes=3,
    lr=0.01
)

model.fit(X, y)

Z = model.forward(X)

probabilities = model.softmax(Z)

predictions = np.argmax(probabilities, axis=1)

accuracy = np.mean(predictions == y)

print("Accuracy:", accuracy)

print("Probabilities:")
print(probabilities[:5])

print("\nTrue labels:")
print(y[:5])

print("\nPredicted labels:")
print(predictions[:5])

end = time()

print("Total Time Taken : ",end-start)
```
